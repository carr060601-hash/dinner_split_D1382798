```python
# 第一組測試：4人、折價100元
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# 資料型別轉換與字串處理
event_name = event_name.strip().lower()
meal = int(meal_amount.replace("NT$", "").replace(",", "").strip())
people = int(people)

# 計算費用
service_fee = meal * service_rate
total = meal + service_fee - discount
per_person = total / people
over_budget = per_person > budget_per_person

# 顯示結果
print("聚餐名稱：", event_name)
print(f"原始餐費：{meal:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total:.2f} 元")
print("分攤人數：", people, "人")
print(f"每人費用：{per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print("是否超過預算：", over_budget)
```
